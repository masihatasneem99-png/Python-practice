```python
"""Creating and printing list"""

l=[10,20,30,40]
print(l)
```

```python
"""Print Even index elements*2 without creating new list"""

l=[10,20,30,40]

for i in range(len(l)):
    if i%2==0:
        l[i]=l[i]*2
print(l)
```

```python
"""Move all the 0's to the right of the list and all the 1's to left without using sort method"""

n=[1,0,1,0,1,0,1,0]

for i in range(len(n)-1):
    if n[i]==0:
        n.remove(n[i])
        n.append(0)
print(n)
        
```

```python
"""Separate all the positive and negative numbers from the list to two other lists."""

n=[10,-1,-2,12,-3,-50]
positive=[]
negative=[]

for i in n:
    if i>=0:
        positive.append(i)
    else:
        negative.append(i)
print(positive)
print(negative)
```

```python
"""Print the prime numbers in the list"""

n=[10,11,12,13,14,17]

for i in n:
    count=0
    for j in range(1,i+1):
        if i%j==0:
            count+=1
    if count==2:
        print(i)
```

```python
"""Find the maximum element from a list"""
n=[10,40,14,36]

max=0

for i in n:
    if i>max:
        max=i

print(max)
```

```python
"""Find the minimum value from a list"""

n=[10,40,14,36]

min=n[0]

for i in n:
    if i<min:
        min=i

print(min)
```

```python
"""Split the file names and print only their extensions"""

files=["demo.py", "day2.html", "loops.py", "db.sql"]

for i in files:
    n=i.split(".")
    print(n[-1])
```

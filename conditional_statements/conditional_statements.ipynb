```python
'''
Check whether a character is vowel or consonant
'''
s="f"
vowels="aeiouAEIOU"

if s in vowels:
    print("s is a vowel")

else:
    print("s is a consonant")
```

```python
'''
Check whether a number is divisible by both 3 and 5
'''

n=15

if n%3==0 and n%5==0:
    print("n is divisible by both 3 and 5")

else:
    print("n is not divisible by both 3 and 5")
```

```python
'''Check whether a character is uppercase, lowercase, digit, or special symbol'''


s="A"

if s.isupper():
    print("s is uppercase")

elif s.islower():
    print("s is lowercase")

elif s.isdigit():
    print("s is a digit")

else:
    print("s is a special symbol")

```

```python
"""Discount based on amount"""

amount=6000

if amount<1000:
    print("no discount")

elif amount<=5000:
    print("10 percent discount")
    amount= amount- ((amount*10)/100)
    print("after discount amount=", amount)

elif amount<=10000:
    print("20 percent discount")
    amount= amount- ((amount*20)/100)
    print("after discount amount=", amount)

elif amount<=30000:
    print("30 percent discount")
    amount= amount- ((amount*30)/100)
    print("after discount amount=", amount)

elif amount<=50000:
    print("40 percent discount")
    amount= amount- ((amount*40)/100)
    print("after discount amount=", amount)

else:
    print("50 percent discount")
    amount= amount- ((amount*50)/100)
    print("after discount amount=", amount)
```

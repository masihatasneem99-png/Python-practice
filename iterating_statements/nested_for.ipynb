```python
"1. ""Palindrome between 100 and 200"""

for i in range(100,200):

    rev=""
    for x in str(i):
        rev=x+rev

    if int(rev)==i:
        print(i)
```

```python
"""2. Pattern 1"""
for row in range(1,6):
    for col in range(1, row+1):
        print(col, end=" ")
    print()
```

```python
"""3. Pattern 2"""
for row in range(1,6):
    for col in range(1,row+1):
        if (row+col)%2==0:
            print("1", end=" ")
        else:
            print("0", end=" ")
    print()
```

```python
"""4. Pattern 3"""

count=1

for i in range(1,5):
    for j in range(1,i+1):
        print(count,end=" ")
        count+=1
    print()
```

```python
"""5. Pattern 4"""
alphabet=65

for i in range(1,6):
    for j in range(1,i+1):
        print(chr(alphabet),end=" ")
        alphabet+=1
    print()
```

```python
"""6. Pattern 5"""

alphabet=65

for i in range(1,6):
    for j in range(1,i+1):
        print(chr(alphabet), end=" ")
    alphabet+=1
    print()
```

```python
"""7. Pattern 6"""

n=65
for i in range(1,6):
    for j in range(1,6):
        print(chr(n), end=" ")
        n+=1

    print()
```

```python
"""8. Pattern 7"""

n=90
for i in range(1,6):
    for j in range(1,6):
        print(chr(n), end=" ")
        n-=1

    print()
```

```python
"""9. Pattern 8"""

for i in range(1,6):
    for j in range(1,i+1):
        if j==1 or i==5 or i==j:
            print(j, end=" ")
        
        else:
            print(" ", end=" ")
    print()
        
```

```python
"""10.pyramid pattern (9)"""

n=6
for i in range(1,n):
    print(" "*(n-i), '* '*i)
```

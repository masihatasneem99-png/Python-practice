```python
"1. ""Palindrome between 100 and 200"""

for i in range(100,200):

    rev=""
    for x in str(i):
        rev=x+rev

    if int(rev)==i:
        print(i)
```

```python
"""2. Pattern 1"""
for row in range(1,6):
    for col in range(1, row+1):
        print(col, end=" ")
    print()
```

```python
"""3. Pattern 2"""
for row in range(1,6):
    for col in range(1,row+1):
        if (row+col)%2==0:
            print("1", end=" ")
        else:
            print("0", end=" ")
    print()
```

```python
"""4. Pattern 3"""

count=1

for i in range(1,5):
    for j in range(1,i+1):
        print(count,end=" ")
        count+=1
    print()
```

```python
"""10.pyramid pattern (9)"""

n=6
for i in range(1,n):
    print(" "*(n-i), '* '*i)
```

```python
"""12. reverse pyramid pattern (11)"""

n=6
for i in range(1,n):
    print("* "*(n-i), " "*i)
```

```python
"""Diamond pattern (12)"""

n=6
for i in range(1,n):
    print(" "*(n-i), '* '*i)
for i in range(n-1,0,-1):
    print(" "*(n-i), "* "*i)
```

```python
"""Pattern 13"""

n=3
for i in range(1,n+1):
    print("* "*i)
for i in range(n-1,0,-1):
    print("* "*i)
```

```python
""" Hollow square pattern (14) """
n=5

for i in range(1,n+1):
    for j in range(1,n+1):
        if i==1 or j==1 or i==n or j==n:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
```

```python
"""A pattern (15)"""

for i in range(1,5):
    for j in range(1,5):
        if i==1 or j==1 or i==3 or j==4:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
```

```python
"""J pattern (16)"""

for row in range(1,6):
    for col in range(1,6):
        if row==1 or col==3:
            print("*", end=" ")
        elif (col in (1,2) and row==5) or (row==4 and col==1):
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
```

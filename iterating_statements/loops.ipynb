```python
"""12. Reverse a number (e.g., 123 → 321). """

n= 123

rev=''

for i in str(n):
    rev= i+ rev

print(rev)
```

```python
"""13. Check whether a number is a palindrome.  """


n=121
rev=''

for i in str(n):
    rev= i +rev

if rev== str(n):
    print("Palindrome")

else:
    print("Not a palindrome")
```

```python
"""16. Check whether a number is an Armstrong number. """

n=153
l=len(str(n))
sum=0

for i in str(n):
    m= int(i) **l
    sum+=m

if sum==n:
    print("Armstrong number")
else:
    print("Not Armstrong number")
```

```python
"""17. Print all prime numbers from 1 to N.  """

n=int(input("Enter a number"))

for i in range(1,n+1):
    count=0
    for x in range(1,i+1):
        if i%x==0:
            count+=1

    if count==2:
        print(i)
```

```python
"""18. Check whether a number is prime or not.
"""

n= int(input("Enter a number:"))
count=0

for i  in range(1,n+1):
    if n%i==0:
        count+=1

if count==2:
    print("Prime")

else:
    print("Not prime")
```

```python
"""19. Find the sum of all prime numbers between 1 and N."""

n=int(input("Enter a number:"))
sum=0

for i in range(1,n+1):
    count=0
    for x in range(1,i+1):
        if i%x==0:
            count+=1
    if count==2:
        sum+=x

print(sum)
```

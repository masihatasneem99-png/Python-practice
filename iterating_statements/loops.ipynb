```python
"2. Print numbers from 1 to N (user input)."

n= int(input("Enter a number:"))

for i in range(1, n+1):
    print(i)
```

```python
"3. Print numbers from N to 1 (reverse order). "

n = int(input("Enter a number:"))

for i in range(n, 0, -1):
    print(i)
```

```python
"4. Print all even numbers from 1 to N.  "

n = int(input("Enter a number:"))

for i in range(0, n+1, 2):
    print(i)
```

```python
"5. Print all odd numbers from 1 to N. "


n = int(input("Enter a number:"))


for i in range(1,n+1,2):
    print(i)
```

```python
"""6. Find the sum of numbers from 1 to N.  
"""

n = int(input("Enter a number:"))
sum=0

for i in range(1, n):
    sum+=i

print(sum)
```

```python
"""7. Find the product (factorial) of a number N. """


n = int(input("Enter a number:"))
product=1

for i in range(1,n+1):
    product*=i

print(product)

```

```python
"""8. Count how many numbers are divisible by 3 between 1 and N. """

n = int(input("Enter a number:"))
count=0

for i in range(1,n):
    if i%3==0:
        count+=1

print(count)
```

```python
"""9. Print the multiplication table of a number N. """

n = int(input("Enter a number:"))

for i in range(1,11):
    print(f"{n} x {i} = {n*i}")
```

```python
"""10. Find the sum of even numbers from 1 to N."""

n = int(input("Enter a number:"))
sum=0

for i in range(0,n,2):
    sum+=i

print(sum)
```

```python
"""11. Find the sum of digits of a number (e.g., 123 → 6).  """

n = int(input("Enter a number:"))
sum=0

for i in str(n):
    sum+= int(i)
print(sum)
```

```python
"""12. Reverse a number (e.g., 123 → 321). """

n= 123

rev=''

for i in str(n):
    rev= i+ rev

print(rev)
```

```python
"""13. Check whether a number is a palindrome.  """


n=121
rev=''

for i in str(n):
    rev= i +rev

if rev== str(n):
    print("Palindrome")

else:
    print("Not a palindrome")
```

```python
"""14. Count the number of digits in a number. """

n=365
m=str(n)
count=0

for i in range(len(m)):
    count+=1

print(count)

```

```python

"""15. Find the largest digit in a number.  """

n=485
largest= 0

for i in str(n):
    m= int(i)

    if m > largest:
        largest=m

print(largest)
```

```python
"""16. Check whether a number is an Armstrong number. """

n=153
l=len(str(n))
sum=0

for i in str(n):
    m= int(i) **l
    sum+=m

if sum==n:
    print("Armstrong number")
else:
    print("Not Armstrong number")
```

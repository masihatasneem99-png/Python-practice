```python
"""Factorial of a number"""

n=5
s=1
fact=1

while s<=n:
    fact*=s
    s+=1
    
print(fact)
```

```python
"""Perfect number: A positive integer that is equal to the sum of its proper divisors(excluding itself)"""

n=10
s=1
sum=0
while s<n:
    if n%s==0:
        sum+=s
    s+=1
if sum==n:
    print("Perfect number")
else:
    print("Not perfect number")
```

```python
"""iterating string using while loop"""

s="python"
n=0

while n<len(s):
    print(s[n])
    n+=1
```

```python
"Print vowels in even index"

n= "strong number"
s=0

while s<len(n):
    if s%2==0:
        if n[s] in "aeiouAEIOU":
            print(n[s])

    s+=1
```

```python
"""Next prime number"""
n=10

while True:
    count=0
    for i in range(1,n+1):
        if n%i==0:
            count+=1
    if count==2:
        print(n)
        break
    n+=1
    
```

```python
"""Tables from 5 to 10"""
n=5

while n<=10:
    s=1
    while s<=10:
        print(f"{n} x {s} = {n*s}")
        s+=1
    n+=1
    print()
```

```python
"""Print odd numbers in a range"""

for x in range(1,10):
    if x%2==0:
        continue
    print(x)
```

```python
"""Print even numbers in a range"""

for x in range(1,10):
    if x%2!=0:
        continue
    print(x)
```

```python
"""Sum of first n numbers"""

sum=0
n=10
s=0
while s<=n:
    sum+=s
    s+=1
print(sum)
```

```python
"""Reverse a number"""

n=1234
rev=0

while n>0:
    digit=n%10
    n=n//10
    rev= rev*10+ digit
print(rev)
```

```python
"""Sum of digits of a number"""

n=1234
sum=0
while n>0:
    digit=n%10
    n=n//10
    sum+=digit
print(sum)
```

```python
"""Product of digits of a number"""

n=1234
product=1
while n>0:
    digit=n%10
    n=n//10
    product*=digit
print(product)
```

```python
"""Largest digit in a number"""

n=907667
largest=0

while n>0:
    digit=n%10
    n=n//10
    if digit>largest:
        largest=digit
print(largest)
```

```python
"""Smallest digit in a number"""

n=243546
smallest=9

while n>0:
    digit= n%10
    n//=10
    if digit<smallest:
        smallest=digit
print(smallest)
```

```python
"""Fibonacci series of n elements"""

n=8
a,b=0,1
while n>0:
    print(a)
    a,b=b,a+b
    n-=1
```

```python
"""Check whether a number is prime or not using while loop """

n=11
s=1
count=0

while s<=n:
    if n%s==0:
        count+=1
    s+=1

if count==2:
    print("Prime")
else:
    print("Not prime")
```

```python
"""Print all prime numbers in a range of 1 to 100 using while loop"""

n=1
while n<100:
    s=1
    count=0
    while s<=n:
        if n%s==0:
            count+=1
        s+=1
    if count==2:
        print(n)
    n+=1
```

```python
"""upper() methods"""

n="a string in python is a sequence"
n=n[0].upper() + n[1::]
print(n)
```

```python
"""Capitalizing the first letter of a sentence"""

n="a string in python is a sequence"
temp=""
for x in range(len(n)):
    if x==0:
        k=n[x].upper()
        temp+=k
    else:
        temp+=n[x]
print(temp)
```

```python
"""Splitting a string without using split() method"""

n="a string in python is a sequence"
l=[]
word=""
for x in n:
    if x !=" ":
        word+=x
    else:
        l.append(word)
        word=""
l.append(word)
print(l)

```

```python
"""Adding '_' before words in a sentence"""

n="StringInPython"
temp=""
for x in n:
    if x>='A' and x<='Z':
        temp=temp+"_"+x
    else:
        temp+=x
print(temp)
```

```python
"""Removing _ before the words"""

n="_String_In_Python"
res=""
word=""
for x in n:
    if x=="_":
        res+=word
        word=""
    else:
        word+=x
res+=word
print(res)
```

```python
"""Removing _ before words (method 2)"""

n="_String_In_Python"
temp=""

for x in n:
    if x!="_":
        temp+=x
print(temp)
```

```python
"""Print the vowels in the prime index of a string"""

n="A string in python is a sequence"

for x in range(len(n)):
    count=0
    for i in range(1,x+1):
        if x%i==0:
            count+=1
    if count==2 and n[x] in 'aeiouAEIOU':
        print(x, n[x])
```

```python
"""Capitalizing first letter of each word"""

n="a string in python is a sequence".split()

s=""

for x in n:
    for i in range(len(x)):
        if i==0:
            if len(x)==1:
                k=x[i].upper()
                s+=k+" "
            else:
                k= x[i].upper()
                s+=k

        elif i==len(x)-1:
            s=s+x[i]+" "
        else:
            s+=x[i]
    
print(s)

```

```python
"""Replace vowels with * in a string"""

n="A string in python"
s=""
for x in n:
    if x in "aeiouAEIOU":
        s+='*'
    else:
        s+=x
print(s)
```

```python
""" Printing the longest word in a string"""

n="string in python is a sequence".split()
largest=0
word=""

for x in n:
    if len(x)>largest:
        word=x
        largest=len(x)
print(word)
```

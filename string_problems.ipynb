```python
"""upper() methods"""

n="a string in python is a sequence"
n=n[0].upper() + n[1::]
print(n)
```

```python
"""Capitalizing the first letter of a sentence"""

n="a string in python is a sequence"
temp=""
for x in range(len(n)):
    if x==0:
        k=n[x].upper()
        temp+=k
    else:
        temp+=n[x]
print(temp)
```

```python
"""Splitting a string without using split() method"""

n="a string in python is a sequence"
l=[]
word=""
for x in n:
    if x !=" ":
        word+=x
    else:
        l.append(word)
        word=""
l.append(word)
print(l)

```

```python
"""Adding '_' before words in a sentence"""

n="StringInPython"
temp=""
for x in n:
    if x>='A' and x<='Z':
        temp=temp+"_"+x
    else:
        temp+=x
print(temp)
```

```python
"""Removing _ before the words"""

n="_String_In_Python"
res=""
word=""
for x in n:
    if x=="_":
        res+=word
        word=""
    else:
        word+=x
res+=word
print(res)
```

```python
"""Removing _ before words (method 2)"""

n="_String_In_Python"
temp=""

for x in n:
    if x!="_":
        temp+=x
print(temp)
```

```python
"""Replace vowels with * in a string"""

n="A string in python"
s=""
for x in n:
    if x in "aeiouAEIOU":
        s+='*'
    else:
        s+=x
print(s)
```

```python
""" Printing the longest word in a string"""

n="string in python is a sequence".split()
largest=0
word=""

for x in n:
    if len(x)>largest:
        word=x
        largest=len(x)
print(word)
```

```python
"""Removing spaces from a string"""

n="string in python is a sequence"

temp=""

for i in n:
    if i ==" ":
        continue
    else:
        temp+=i
print(temp)
```

```python
"""Separating elements in an emial address"""

n="masiha123@gmail.com"
alp=""
nums=""
special=""
for i in n:
    if i.isdigit():
        nums+=i
    elif i.isalpha():
        alp+=i
    else:
        special+=i
print(alp)
print(nums)
print(special)
```

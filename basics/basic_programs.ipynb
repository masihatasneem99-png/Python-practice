```python
'''
Take a user's full name as input and print their initials in uppercase.

'''

name=input("Enter your full name:")

initial=""

name=name.split(" ")

for i in name:
    initial+= i[0].upper()

print(initial)
```

```python
'''
Given a string, extract and print the first half of the string.

'''

example="my name is masiha"

for x in range(int(len(example)/2)):
    print(example[x])
```

```python
'''
Take a sentence as input 
and print it in reverse order without using built-in functions
'''

sentence=input("enter a sentence:")

print(sentence[::-1])
```

```python
'''
Check if a given string contains only numeric characters'''

characters=input("Enter a string")
numeric= False
for n in characters:
    if n.isdigit() or n==" ":
        numeric= True
    else:
        numeric= False
        break

print(numeric)
```

```python
"""Check if a number is divisible by 7."""

num=int(input("enter a number:"))

if num%7==0:
    print("divisible by 7")

else:
    print("not divisible by 7")

```
